```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

maison = pd.read_parquet("house_prices.parquet")
maison_numerique = maison.select_dtypes(include='number')
correlations = maison_numerique.corrwith(maison_numerique['price']).sort_values(ascending=False)

plt.figure(figsize=(12,6))
sns.heatmap(correlations.to_frame(), annot=True,cmap='coolwarm' )
plt.title("Influence des colonnes numériques sur le prix des maisons")
plt.show()
```
