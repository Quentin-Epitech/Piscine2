```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

lecture = pd.read_csv("weather.csv")

ville = 'Sydney'
ville_lecture = lecture[lecture['city_name'] == ville]

plt.figure(figsize=(10,6))
sns.scatterplot(data=ville_lecture, x='min_temp_c', y='max_temp_c', hue='season')
plt.title(f"Répartition des températures minimales et maximales par saison à {ville}")
plt.xlabel("Température minimale en degrés")
plt.ylabel("Température maximale en degrés")
plt.legend(title='Saison')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


sns.displot(ville_lecture, x="min_temp_c", hue="season", kind="kde")
plt.title(f"Distribution des températures minimales par saison")
plt.show()

sns.displot(ville_lecture, x="max_temp_c", hue="season", kind="kde")
plt.title(f"Distribution des températures maximales par saison")
plt.show()

```

```python
sns.displot(ville_lecture, x="avg_temp_c", hue="season", kind="hist", multiple="stack")
plt.title(f"Distribution de la température moyenne par saison (hist)")
plt.show()

```
