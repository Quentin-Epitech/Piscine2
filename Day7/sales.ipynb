```python
import pandas as pd
import matplotlib.pyplot as plt


lecture = pd.read_csv("sales.csv", na_values=[])

repartition = lecture.groupby("region")["total_price"].sum()

plt.figure(figsize=(8,8))
plt.pie(repartition, labels=repartition.index, autopct='%1.1f%%',startangle=90)
plt.title("Répartition des ventes par région")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

lecture = pd.read_csv("sales.csv", na_values=[])

repartition = lecture.groupby("category")["total_price"].sum()

plt.figure(figsize=(8,8))
plt.pie(repartition,labels=repartition.index,autopct='%1.1f%%',startangle=90)
plt.title("Répartition des ventes par catégorie de produit")
plt.show()
```

La région **EMEA** dominent énormement le marché(**78.9%**),et la répartition des ventes,les **voitures classiques** dominent elles aussi le marché avec **48.8%**.
Pour conclure,nous pouvons dire que le marché est très hétérogène .

```python
sns.displot(lecture, x="total_price", kde=True)
plt.title("Distribution du prix total de vente")
plt.show()
```

```python
sns.displot(lecture, x="total_price", col="category", kde=True)
plt.suptitle("Distribution du prix total par catégorie", y=1.05)
plt.show()
```

```python
sns.displot(lecture, x="total_price", col="region", hue="category", kde=True)
plt.suptitle("Distribution du prix total par catégorie et par région", y=1.05)
plt.show()
```

```python
sns.displot(lecture, x="total_price", col="region", kde=True)
plt.suptitle("Distribution du prix total par région", y=1.05)
plt.show()
```

```python
sns.displot(lecture, x="total_price", col="dealsize", kde=True)
plt.suptitle("Distribution du prix total par dealsize", y=1.05)
plt.show()
```

```python
sns.displot(lecture, x="total_price", col="region", hue="dealsize", kde=True)
plt.suptitle("Distribution du prix total par dealsize et par région", y=1.05)
plt.show()
```
